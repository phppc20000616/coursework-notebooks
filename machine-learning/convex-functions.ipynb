```python
# (d)
C = np.array([[2,1],[1,3]])
b = np.array([1,1])

def g_d(w1, w2):
    w = np.array([w1, w2])
    return 0.5 * w.T @ C @ w + b.T @ w

# 网格
w1 = np.linspace(-2, 1, 200)
w2 = np.linspace(-2, 1, 200)
W1, W2 = np.meshgrid(w1, w2)
G = np.zeros_like(W1)
for i in range(W1.shape[0]):
    for j in range(W1.shape[1]):
        G[i,j] = g_d(W1[i,j], W2[i,j])

# stationary point
w_star = -np.linalg.inv(C) @ b

plt.figure(figsize=(6,5))
CS = plt.contour(W1, W2, G, levels=20)
plt.clabel(CS, inline=True, fontsize=8)
plt.scatter([w_star[0]], [w_star[1]], color="r", s=60, label="stationary point")
plt.title("(d) g(w) quadratic form")
plt.xlabel("w1"); plt.ylabel("w2")
plt.legend(); plt.grid(True)
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

# Define the cost function and its gradient
def g(w):
  """Cost function g(w) = w^T * w"""
  # CORRECTED LINE: np.sum(w**2) returns a scalar value.
  return np.sum(w**2)

def grad(w):
  """Gradient of the cost function, nabla g(w) = 2w"""
  return 2*w

# Gradient descent algorithm
def gradient_descent(alpha, max_its, w_initial):
  """
  Runs gradient descent for a given steplength (alpha),
  number of iterations (max_its), and initial point (w_initial).
  """
  w = w_initial.copy()
  # The list will now be populated with scalars, which is what matplotlib expects.
  cost_history = [g(w)]
  
  for k in range(max_its):
    # Calculate the gradient
    gradient = grad(w)
    
    # Take a gradient descent step
    w = w - alpha * gradient
    
    # Store the cost
    cost_history.append(g(w))
    
  return cost_history

# --- Setup and Run ---
# Initial parameters
N = 10
w0 = 10 * np.ones((N, 1))
max_iterations = 100
alphas = [0.001, 0.1, 1]

# Run gradient descent for each alpha
cost_history_1 = gradient_descent(alphas[0], max_iterations, w0)
cost_history_2 = gradient_descent(alphas[1], max_iterations, w0)
cost_history_3 = gradient_descent(alphas[2], max_iterations, w0)

# --- Plotting ---
plt.figure(figsize=(12, 7))
plt.plot(cost_history_1, label=fr'$\alpha = {alphas[0]}$')
plt.plot(cost_history_2, label=fr'$\alpha = {alphas[1]}$')
plt.plot(cost_history_3, label=fr'$\alpha = {alphas[2]}$', linestyle='--')

plt.title('Gradient Descent Cost History')
plt.xlabel('Iteration')
plt.ylabel(r'Cost $g(\mathbf{w}) = \mathbf{w}^T\mathbf{w}$')
plt.legend()
plt.grid(True)
plt.show()
```
